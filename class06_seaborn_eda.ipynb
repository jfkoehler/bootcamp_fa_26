````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
````

## 1. Retail pricing: the diamonds dataset

Each row represents a diamond offered for sale. The dataset contains physical characteristics, quality categories, and price.


````python
DIAMONDS_URL = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv'

diamonds = pd.read_csv(DIAMONDS_URL)
diamonds.head()
````

````python
diamonds.info()
````

Important columns include:

| Column | Meaning |
|---|---|
| `carat` | Diamond weight |
| `cut` | Cut-quality category |
| `color` | Color grade |
| `clarity` | Clarity grade |
| `price` | Price in U.S. dollars |

The quality categories are labels, while `carat` and `price` are numerical measurements.


## 2. The basic Seaborn pattern

Most Seaborn commands specify `data`, `x`, `y`, and sometimes `hue`. We store the plot axes in `ax` so we can add a title and labels.

### Count plot: how many observations belong to each category?


````python
ax = sns.countplot(data=diamonds, x='cut')
ax.set_title('Number of Diamonds by Cut Quality')
ax.set_xlabel('Cut')
ax.set_ylabel('Number of Diamonds')
````

A count plot counts rows; it does not display an average price. What business question could a retailer answer with this plot?


## 3. Compare a numerical variable across categories

### Box plot: compare distributions

Prices contain large values and outliers. A box plot helps compare the center, spread, and unusual values across cut categories.


````python
ax = sns.boxplot(data=diamonds, x='cut', y='price')
ax.set_title('Diamond Price Distribution by Cut')
ax.set_xlabel('Cut')
ax.set_ylabel('Price ($)')
````

**Discuss:** Does a better cut category automatically produce a higher price? What other product characteristics could affect price?


#### The `hue` argument

Using `hue` splits the plot by a category and allows a third dimension to enter your visualization. 

````python
ideal_cut = diamonds.loc[diamonds['cut'].isin(['Ideal', 'Premium'])]
````

````python
plt.figure(figsize = (10, 5))
ax = sns.boxplot(data=ideal_cut, x='cut', y='price', hue = 'color')
ax.set_title('Diamond Price Distribution by Cut')
ax.set_xlabel('Cut')
ax.set_ylabel('Price ($)')
ax.legend(loc = 'upper right')
````

### Scatterplot: compare two numerical variables

The full dataset contains more than 50,000 rows, so we take a reproducible sample to keep the plot readable.


````python
diamond_sample = diamonds.sample(n=3000, random_state=42)

ax = sns.scatterplot(
    data=diamond_sample,
    x='carat',
    y='price',
    hue='cut',
    alpha=0.65
)
ax.set_title('Diamond Price and Carat Weight')
ax.set_xlabel('Carat')
ax.set_ylabel('Price ($)')
plt.show()
````

**Discuss:** What relationship appears between carat and price? Why does the plot not isolate the effect of cut quality?
