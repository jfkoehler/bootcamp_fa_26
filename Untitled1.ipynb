```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
```

```python
numeric = titanic.select_dtypes(['float', 'int'])
```

```python
numeric.head()
```

```python
melted = pd.melt(numeric, 'survived')
```

```python
melted.head()
```

```python
sns.boxplot(melted, x = 'variable', y = 'value', hue = 'survived', )
plt.ylim(0,200)
```

```python
sns.scatterplot(numeric, x = 'fare', y = 'survived', hue = 'survived')
```

```python
def line(m,b=0, x=None):
    return m*x + b
```

```python
x = numeric['fare']
```

```python
line(.2/100, b = 0, x = x)
```

```python
sns.scatterplot(numeric, x = 'fare', y = 'survived', hue = 'survived')
plt.plot(x, line(.2/100, b = 0, x = x), '-r');
```

```python
def sigmoid(x):
    return np.exp(-x)/(1 + np.exp(-x))
```

```python
sns.scatterplot(numeric, x = 'fare', y = 'survived', hue = 'survived')
```

```python
sigmoid(line(.2/1000, b = 0, x = x))
```

```python
x = np.linspace(0, 500, 1000)
```

```python
from ipywidgets import interact
import ipywidgets as widgets
```

```python
def sigmoid(m, b):
    return plt.plot(x, 1/(1 + np.exp(-m*x - b)))
```

```python
interact(sigmoid,
         m = widgets.FloatSlider(value = 0, min = -5, max = 5),
         b = widgets.FloatSlider(value = 0, min = 5, max = 50),
         )
```
