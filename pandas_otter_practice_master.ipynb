## Question 1: Load a CSV from a URL

The variable `data_url` contains a direct link to a CSV file. Use `pd.read_csv()` to load it into a DataFrame named `titanic`.


```python
data_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

titanic = pd.read_csv(data_url)  # SOLUTION
titanic.head()
```

```python
def test_q1_public(titanic, pd):
    assert isinstance(titanic, pd.DataFrame), "`titanic` should be a pandas DataFrame."
    assert titanic.shape == (891, 15), "Check that you loaded the CSV from `data_url`."
```

```python
""" # BEGIN TEST CONFIG
hidden: true
""" # END TEST CONFIG
def test_q1_hidden(titanic):
    required = {"survived", "pclass", "sex", "age", "fare"}
    assert required.issubset(titanic.columns)
    assert titanic.index.equals(titanic.reset_index(drop=True).index)
```

## Question 2: Examine missing values

Create a Series named `missing_counts` containing the number of missing values in every column. Use `.isna()` and `.sum()`.


```python
missing_counts = titanic.isna().sum()  # SOLUTION
missing_counts
```

```python
def test_q2_public(missing_counts, titanic, pd):
    assert isinstance(missing_counts, pd.Series), "`missing_counts` should be a Series."
    pd.testing.assert_series_equal(missing_counts, titanic.isna().sum())
```

```python
""" # BEGIN TEST CONFIG
hidden: true
""" # END TEST CONFIG
def test_q2_hidden(missing_counts):
    assert int(missing_counts["age"]) == 177
    assert int(missing_counts["deck"]) == 688
    assert int(missing_counts["embarked"]) == 2
```

## Question 3: Select columns

Create a DataFrame named `passenger_info` containing these five columns in this exact order:

`survived`, `pclass`, `sex`, `age`, `fare`


```python
passenger_info = titanic[["survived", "pclass", "sex", "age", "fare"]]  # SOLUTION
passenger_info.head()
```

```python
def test_q3_public(passenger_info, titanic, pd):
    assert isinstance(passenger_info, pd.DataFrame), "`passenger_info` should be a DataFrame."
    expected = titanic[["survived", "pclass", "sex", "age", "fare"]]
    pd.testing.assert_frame_equal(passenger_info, expected)
```

```python
""" # BEGIN TEST CONFIG
hidden: true
""" # END TEST CONFIG
def test_q3_hidden(passenger_info):
    assert passenger_info.columns.tolist() == ["survived", "pclass", "sex", "age", "fare"]
    assert passenger_info.shape == (891, 5)
```

## Question 4: Filter rows

Create a DataFrame named `adult_first_class` containing passengers who were in first class (`pclass == 1`) **and** were at least 18 years old. Keep all columns.


```python
adult_first_class = titanic[(titanic["pclass"] == 1) & (titanic["age"] >= 18)]  # SOLUTION
adult_first_class.head()
```

```python
def test_q4_public(adult_first_class, titanic, pd):
    assert isinstance(adult_first_class, pd.DataFrame), "`adult_first_class` should be a DataFrame."
    expected = titanic[(titanic["pclass"] == 1) & (titanic["age"] >= 18)]
    pd.testing.assert_frame_equal(adult_first_class, expected)
```

```python
""" # BEGIN TEST CONFIG
hidden: true
""" # END TEST CONFIG
def test_q4_hidden(adult_first_class, titanic):
    assert adult_first_class.columns.tolist() == titanic.columns.tolist()
    assert (adult_first_class["pclass"] == 1).all()
    assert (adult_first_class["age"] >= 18).all()
```

## Question 5: Use `groupby`

Create a Series named `survival_rate_by_class` containing the mean of `survived` for each value of `pclass`. The index should contain the passenger classes.


```python
survival_rate_by_class = titanic.groupby("pclass")["survived"].mean()  # SOLUTION
survival_rate_by_class
```

```python
def test_q5_public(survival_rate_by_class, titanic, pd):
    assert isinstance(survival_rate_by_class, pd.Series), "`survival_rate_by_class` should be a Series."
    expected = titanic.groupby("pclass")["survived"].mean()
    pd.testing.assert_series_equal(survival_rate_by_class, expected)
```

```python
""" # BEGIN TEST CONFIG
hidden: true
""" # END TEST CONFIG
def test_q5_hidden(survival_rate_by_class):
    assert survival_rate_by_class.index.tolist() == [1, 2, 3]
    assert abs(float(survival_rate_by_class.loc[1]) - 0.6296296296296297) < 1e-12
    assert abs(float(survival_rate_by_class.loc[3]) - 0.24236252545824846) < 1e-12
```
