## Part IV: Real-world business case — Zillow Offers

### Case reading: When a prediction becomes a business decision

Zillow is widely known for publishing online estimates of home values. In 2018, the company expanded into a different kind of business through **Zillow Offers**. Instead of only displaying information about homes, Zillow began buying homes directly from owners, preparing them for resale, and selling them. This is often called **iBuying**.

The business depended on data at several stages. Zillow had to estimate a home's current value, predict what it could sell for later, estimate renovation and transaction costs, and decide how much risk to accept. A prediction error on a website might inconvenience a user. A prediction error in Zillow Offers could cause the company to buy a physical asset for more than it could later recover. Each purchase also committed money, employees, contractors, and operational capacity.

The environment became especially difficult during the rapidly changing housing market of 2021. Historical relationships could not guarantee that prices, buyer demand, repair costs, or the time required to resell a home would remain stable. Zillow also faced renovation and resale capacity constraints. This meant that the business problem was larger than producing an accurate home-value estimate: the company had to coordinate forecasts with thousands of individual purchasing decisions and with the practical ability to renovate and resell its inventory.

On November 2, 2021, Zillow announced that it would wind down Zillow Offers. CEO Rich Barton said that the unpredictability of forecasting home prices was greater than the company had anticipated and that continuing to scale the operation would create too much earnings and balance-sheet volatility. Zillow reported an inventory write-down of approximately **\$304 million** in the third quarter because it had purchased homes at prices above its updated estimates of their future selling prices. It also expected an additional **\\$240 million to \$265 million** in losses, primarily related to homes it expected to purchase in the fourth quarter. The wind-down was expected to reduce Zillow's workforce by approximately **25%**.

The case does not show that data or prediction models are useless. It shows that a model must be evaluated inside the business system where its predictions become actions. Decision makers must consider the size and direction of errors, changing market conditions, operational bottlenecks, exposure created by scaling, and the cost of being wrong. A model can be informative while still being unsuitable for an automated or rapidly expanding decision process.
