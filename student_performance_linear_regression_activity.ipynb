## 6. Evaluate the model with `.score()`

For `LinearRegression`, `.score()` returns the coefficient of determination, usually written as $R^2$.

- A value near 1 means the model explains a large share of the variation in the response.
- A value near 0 means the model explains little more than simply predicting the mean.
- A score can be negative when predictions are particularly poor.
