## 2. The shell and the file system

A **shell** is a text interface for giving instructions to your computer. Most commands follow this pattern:

```text
command  options  target
ls       -la      class02_sandbox
```

The shell is literal: capitalization, spaces, punctuation, and the current folder all matter. Read error messages before trying a different command. Lines beginning with `#` are comments.

### Where am I? What is here?

```bash
pwd          # print working directory
ls           # list visible files and folders
ls -la       # long listing, including hidden items
```


````python
%%bash
pwd
ls -la
````

### Paths and navigation

A path describes a location. An **absolute path** starts at the root of the file system. A **relative path** starts from your current directory.

| Expression | Meaning |
| --- | --- |
| `.` | current directory |
| `..` | parent directory |
| `~` | your home directory |
| `/` | root directory, or a separator within a path |

Navigate with `cd`:

```bash
cd folder_name       # enter a folder
cd ..                # move up one level
cd ~                 # go to your home directory
cd "My Projects"   # quote a path containing spaces
```

Use the **Tab** key to complete paths and reduce typing errors. Use the **Up Arrow** to recall earlier commands.

### Inspect text files

```bash
cat file.txt             # print a short file
head file.csv            # show the first 10 lines
tail file.csv            # show the last 10 lines
wc -l file.csv           # count lines
grep pattern file.csv    # show lines matching a pattern
```

Redirection and pipes connect commands:

- `>` writes output to a file, replacing its contents.
- `>>` appends output to a file.
- `|` sends one command's output into the next command.


````python
%%bash
echo 'First two lines:'
head -n 2 class02_sandbox/data/sales.csv
echo
echo 'Number of data rows:'
tail -n +2 class02_sandbox/data/sales.csv | wc -l
echo
echo 'Rows containing Downtown:'
grep 'Downtown' class02_sandbox/data/sales.csv
````

### Copy, move, and remove

```bash
cp source.csv copy.csv             # copy a file
mv old_name.csv new_name.csv       # move or rename a file
rm unwanted_file.csv               # permanently remove a file
rm -r unwanted_folder              # permanently remove a folder and its contents
```

> **Important:** Terminal deletion usually bypasses the Trash. Before using `rm`, run `pwd` and `ls` and check the target carefully. Never run a destructive command you do not understand.

## 3. Git and GitHub

**Git** records versions of a project on your computer. **GitHub** hosts a remote copy that can be shared and synchronized. They are related, but they are not the same thing.

| Term | Meaning |
| --- | --- |
| Working directory | The files currently on your computer |
| Repository | A project whose history Git tracks |
| Commit | A named snapshot of staged changes |
| Remote | A connected copy of the repository, often on GitHub |
| Clone | A local copy of a remote repository, including its history |

### One-time configuration

Run these commands in Git Bash or Terminal, replacing the examples with the name and email associated with your GitHub account:

```bash
git config --global user.name "Ada Lovelace"
git config --global user.email "ada@example.com"
git config --global init.defaultBranch main
git config --global --list
```


### Everyday Git workflow

Clone the course repository once. Replace the placeholders with the information provided in class:

```bash
git clone <COURSE_REPOSITORY_URL>
cd <COURSE_REPOSITORY_FOLDER>
```

At the beginning of a work session:

```bash
git status
git pull
```

After making a meaningful change:

```bash
git status
git diff
git add path/to/file.ipynb
git status
git commit -m "Complete technology setup exercise"
git push
```

Use a specific commit message that explains what changed. Do not type `git add .` until you have checked `git status` and understand every file that will be included.

### When something goes wrong

Stop and inspect before experimenting:

```bash
pwd
git status
git log --oneline -5
```

Copy the complete error message when asking for help. Avoid force pushes, resets, or deleting `.git`; those actions can discard work.

## Command reference

| Command | Purpose |
| --- | --- |
| `pwd` | print the current directory |
| `ls -la` | list files with details, including hidden files |
| `cd path` | change directory |
| `mkdir -p path` | create a directory and any needed parent directories |
| `touch file` | create an empty file or update its timestamp |
| `cp source target` | copy a file |
| `mv source target` | move or rename a file |
| `head -n 5 file` | display the first five lines |
| `tail -n 5 file` | display the last five lines |
| `wc -l file` | count lines |
| `grep pattern file` | print lines matching a pattern |
| `git status` | show the repository's current state |
| `git diff` | show unstaged changes |
| `git add file` | stage a specific file |
| `git commit -m "message"` | record staged changes |
| `git pull` | retrieve and integrate remote changes |
| `git push` | send local commits to the remote |
