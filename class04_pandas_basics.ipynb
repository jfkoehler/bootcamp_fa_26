## 3. Filtering with conditions

A comparison produces one Boolean value per row. Use the resulting Boolean Series inside `.loc[]` to retain matching rows.

```python
positive_balance = bank["balance"] > 1000
positive_balance.head()
```

```python
bank.loc[positive_balance, ["age", "job", "balance", "y"]].head()
```

Combine conditions with:

- `&` for **and**;
- `|` for **or**; and
- `~` for **not**.

Place each comparison in parentheses. Do not use Python's `and` or `or` with pandas Series.

```python
priority_clients = bank.loc[
    (bank["balance"] > 1000)
    & (bank["housing"] == "no")
    & (bank["loan"] == "no"),
    ["age", "job", "balance", "housing", "loan", "y"]
]
priority_clients.head()
```

```python
print("Matching clients:", len(priority_clients))
priority_clients["y"].value_counts()
```

## 4. Grouping and aggregation

`groupby()` follows a split–apply–combine pattern:

1. split rows into categories;
2. apply a calculation to each category; and
3. combine the results into a new table.

```python
bank.groupby("job")["balance"].mean()
```

```python
bank.groupby("job")["balance"].mean().sort_values(ascending=False)
```

To calculate a response rate, first convert `yes` and `no` into Boolean values. In Python, `True` behaves like 1 and `False` like 0, so the mean is the proportion of `True` values.

```python
bank_analysis = bank.assign(subscribed=bank["y"].eq("yes"))

job_summary = (
    bank_analysis.groupby("job")
    .agg(
        clients=("y", "size"),
        average_balance=("balance", "mean"),
        subscription_rate=("subscribed", "mean"),
    )
    .sort_values("subscription_rate", ascending=False)
)
job_summary.round(3)
```
