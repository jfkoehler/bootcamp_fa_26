# Data Bootcamp
## Fall 2026

Data Bootcamp is about the nuts and bolts of data analysis. You will learn how to analyze economic and business data and enough computer programming to work with data efficiently.
We will use Python, a popular high-level programming language widely used in finance, consulting, technology, and other parts of the business world. “High-level” means that much of the underlying work is handled by the language, but Python remains a serious language with extensive capabilities.

“Data analysis” means using data to describe patterns, answer questions, and communicate useful insights. “Bootcamp” is a reminder that expertise takes work. The effort is worthwhile: you will build practical skills that are valuable to current and future employers and learn to do things more effectively than you could with spreadsheets alone.

## Requirements



There are no prerequisites. We welcome students with no prior programming experience and have designed the course with them in mind. However, this is a demanding, fast-paced course. The first class introduces the course and how to frame a business-ready data challenge. The second class establishes the course technology workflow with Git Bash or Terminal, Unix commands, Git, and GitHub. We will begin working with Python and pandas in the third class and build quickly toward statistical modeling, machine learning, and neural networks. Success requires consistent attendance, substantial practice between meetings, and the patience to diagnose programs that do not work—a regular occurrence even for experts.

## Deliverables and Grades

The course uses the CRISP-DM cycle to connect business understanding, data understanding, preparation, modeling, evaluation, and deployment. After an opening class on business problem framing and a hands-on technology workshop, Python fundamentals and pandas begin in the third class. The course then moves quickly through data acquisition, visualization, statistical reasoning, predictive modeling, text and time-series analysis, and neural networks. Throughout the semester, you will complete short quizzes, code-practice assignments, mini-projects, and presentations. The course culminates in a final project that you can share with potential employers as evidence of your quantitative and communication skills.

Every project will begin with a concise CRISP-DM brief establishing the business objective, what the available data represents and omits, and a feasible analytical direction.

Graded work includes:

- **Code Practice:** Regular assignments will help you build fluency, reinforce material from class, and identify questions to bring to the next meeting.
- **In-Class Quizzes:** Short, low-stakes quizzes will be completed in class through nbgrader on NYU JupyterHub. They will focus on essential concepts and small coding tasks drawn from recent class meetings and assignments.
- **Mini-Projects and Presentations:** Frequent mini-projects will ask you to apply several skills to a meaningful dataset or question. Selected projects will include short presentations, demonstrations, or peer-feedback sessions so that you practice explaining both your process and your findings.
- **Final Project and Showcase:** You will develop a substantial final project through a proposal, checkpoints, peer feedback, and a final presentation. The finished project should be a professional piece of data collection, analysis, and communication that you can share with potential employers. Strong projects depend more on sustained thought and effort than on previous programming experience.

Due dates will be posted on the course Brightspace site.

Dates are not negotiable. Anything handed in late will get a grade of zero.

All your work should be clean and professional. Your grade depends on it.

Final grades will be computed as follows:

## Recommended Work Habits

Python is not something you can learn only by reading a book and attending lectures. You need to write programs—the more the better—to understand how they work. Think about learning basketball or soccer: reading and listening are not enough; you need to practice. We will do substantial programming in class, but it is essential that you follow up outside class.
Here's how.


After each topic, we recommend that you write, review, and practice:

- Write: Shortly after class, write down everything you remember without looking at your notes or the book. Note things you don't understand.
- Review: Read the relevant section of the book. Fill in the gaps. Ask for help with anything you still don't understand.
- **Practice:** Complete the associated code practice soon after class. Re-create examples without copying, experiment with small changes, and use quiz feedback to decide what to revisit.
- **Apply:** Treat each mini-project as a chance to connect techniques rather than demonstrate one isolated skill. Start with a clear question, make deliberate choices, and explain what the results do—and do not—show.
- **Present:** Practice describing your question, data, approach, and findings to a nontechnical audience. Clear communication is part of data analysis, not an optional addition to it.

We also recommend you practice coding whenever you have the chance. Start small. Write short programs to do anything that crosses your mind. Use Python to do things you would ordinarily do in Excel. Try doing assignments from other courses in Python. At first this will be more work than doing it by hand or in Excel, but once you have some experience it will typically be easier in Python. Even if that's not the case, the practice will expand your skill set.
