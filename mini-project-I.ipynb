# Mini Group Project: Peer-Feedback Rubric

Use this rubric while another group presents. Your purpose is to help the group make a focused revision before the final presentation -- not simply to assign a score.

## Four-point scale

| Criterion | 4  Strong | 3  Effective | 2  Developing | 1  Needs attention |
|---|---|---|---|---|
| **Question and significance** | The research question is focused, answerable with the available data, and its importance is compellingly explained. | The question is understandable and answerable; its importance is mostly clear. | The question is broad, somewhat unclear, or only partly matched to the available data. | The question is missing, confusing, or cannot be addressed with the presented data. |
| **Data and analytical choices** | The dataset, source, unit of observation, preparation, and analytical choices are explained clearly and appropriately. | The main data and methods are explained, with only minor gaps. | Some important data or method choices are unclear, unexplained, or weakly connected to the question. | The audience cannot determine what the data represent or how the analysis was produced. |
| **Visual evidence** | Visualizations are readable, accurately labeled, well chosen, and directly support the presentations claims. | Visualizations are generally clear and relevant, with minor design or labeling issues. | Visualizations contain distracting design choices, incomplete labels, or a weak connection to the claims. | Visualizations are missing, difficult to read, misleading, or unrelated to the question. |
| **Interpretation and limitations** | Findings are specific and supported by evidence; limitations and uncertainty are addressed thoughtfully without causal overclaiming. | Findings are supported and at least one meaningful limitation is acknowledged. | Findings are vague or partially supported; limitations receive little attention. | Claims are unsupported or overstated, and important limitations are ignored. |
| **Presentation and organization** | The presentation has a clear progression, stays within time, involves the group effectively, and ends with a memorable conclusion. | The presentation is understandable and reasonably organized, with generally balanced participation. | Organization, timing, transitions, or participation make parts of the presentation difficult to follow. | The presentation lacks a clear structure, substantially exceeds or falls short of the time, or depends almost entirely on one speaker. |

## Required written feedback

Your feedback should include all five elements:

1. **Strongest element:** Identify one specific choice that helped you understand or trust the analysis.
2. **Evidence question:** Identify one claim that needs stronger evidence, context, or explanation.
3. **Audience question:** Ask one genuine question that a general audience might have.
4. **Visualization revision:** Recommend one concrete improvement to a graph, table, slide, or notebook display.
5. **Priority revision:** Name the single most important change the group should make before presenting again.

## Feedback guidelines

- Refer to a specific moment, claim, or visual rather than saying only good or confusing
- Separate the clarity of the presentation from whether you personally agree with its topic or conclusion.
- Phrase criticism as an actionable suggestion.
- Do not demand a larger or more complicated project; recommend a revision appropriate to the assignments scope.
- Use the numerical ratings to support the written feedback, not replace it.

## Suggested rehearsal timing

- Presentation: 4 minutes
- Quiet rubric ratings: 1 minute
- Spoken peer feedback: 2 minutes
- Groups switch roles and repeat
- Final revision notes: 3-4 minutes
