## 1. Variables and basic data types

A **variable** is a name that refers to a value. Assignment uses a single equals sign. Variable names should be descriptive, cannot contain spaces, and cannot begin with a number.

````python
store_name = "Downtown"       # string: text
transactions = 118             # integer: whole number
revenue = 4200.50               # float: decimal number
promotion_active = True         # Boolean: True or False

print(store_name)
print(transactions, revenue, promotion_active)
````

Use `type()` to inspect a value. Python is case-sensitive, so `True` is valid but `true` is not.

````python
print(type(store_name))
print(type(transactions))
print(type(revenue))
print(type(promotion_active))
````

### Expressions and type conversion

Variables can be combined in expressions. Conversions such as `int()`, `float()`, and `str()` create a value of a different type when the conversion is valid.

````python
average_sale = revenue / transactions
summary = store_name + " average sale: $" + str(round(average_sale, 2))

print(average_sale)
print(summary)
````

#### Aside: f-strings

In contemporary Python, we use `f-strings` to inject variables into a printed statement.  Prefacing a string definition with a `f` creates an f-string.

```python

```

## 2. Collections

Collections store several values in one object. Today we will use three common collection types.

| `list` | Yes | Yes | A sequence of related items |
| `tuple` | Yes | No | A fixed group of values |
| `dict` | In insertion order | Yes | Key-value pairs describing an object |

## 3. Introduction to Git

Git is a tool for tracking changes to files. A folder tracked by Git is called a **repository**.

Today, we will work only with a repository on your computer. We will not connect it to GitHub or use any remote commands.

Git commands should be entered in Git Bash or Terminal, not in a Python notebook cell.

### Create a local repository

Create a practice folder, enter it, and initialize a Git repository:

```bash
mkdir class03_git_practice
cd class03_git_practice
git init
```

The `git init` command tells Git to begin tracking changes in this folder.

Check the repository’s current state:

```bash
git status
```

### Create a Python file

Create a file named `coffee_shop.py`:

```bash
touch coffee_shop.py
```

Open the file in Visual Studio Code:

```bash
code coffee_shop.py
```

Add the following Python code:

```python
store_name = "Downtown"
transactions = 118
revenue = 978.00
promotion_active = False

menu_items = ["coffee", "tea", "pastry"]

store = {
    "name": store_name,
    "transactions": transactions,
    "revenue": revenue,
}

print("Store:", store_name)
print("Transactions:", transactions)
print("Revenue:", revenue)
print("Menu items:", menu_items)
print("Store information:", store)
```

Save the file and run it:

```bash
python coffee_shop.py
```

Depending on your installation, you may need to use:

```bash
python3 coffee_shop.py
```

### Inspect the repository

Check the repository after creating the file:

```bash
git status
```

The file should appear as **untracked** because Git has not yet been told to include it in a commit.

Because this is a new, untracked file, a regular `git diff` may not display its contents.

### Stage the file

Use `git add` to prepare the file for a commit:

```bash
git add coffee_shop.py
git status
```

The file is now in the **staging area**. The staging area contains the changes that will be included in the next commit.

Inspect the staged file:

```bash
git diff --staged
```

The output shows the complete contents of the new Python file because the entire file will be included in the first commit.

### Create the first commit

A **commit** records a snapshot of the staged changes.

```bash
git commit -m "Create coffee shop program"
```

Check the repository again:

```bash
git status
```

If the commit was successful, Git should report that there is nothing to commit and the working tree is clean.

View the repository’s commit history:

```bash
git log --oneline
```

### Modify the Python file

Open `coffee_shop.py` again:

```bash
code coffee_shop.py
```

Add a new item to the `menu_items` list:

```python
menu_items.append("sandwich")
```

Then calculate the average revenue per transaction:

```python
average_sale = revenue / transactions

print("Updated menu:", menu_items)
print("Average sale:", round(average_sale, 2))
```

Save the file and run it again:

```bash
python coffee_shop.py
```

### Inspect the changes with `git diff`

Check the repository:

```bash
git status
```

Git should report that `coffee_shop.py` has been modified.

Inspect the changes:

```bash
git diff
```

The output shows the difference between the last committed version and the current version.

In the output:

- lines beginning with `-` were removed;
- lines beginning with `+` were added; and
- unchanged lines provide context.

### Stage and inspect the changes

Stage the updated file:

```bash
git add coffee_shop.py
git status
```

After staging the file, a regular `git diff` will no longer show the changes. The changes have moved from the working directory to the staging area.

Inspect the staged changes:

```bash
git diff --staged
```

### Create the second commit

Record the updated version:

```bash
git commit -m "Add menu item and average sale"
```

View the repository’s history:

```bash
git log --oneline
```

You should now see two commits:

1. the initial coffee-shop program; and
2. the update that added a menu item and calculated the average sale.

### The basic Git workflow

The workflow we followed was:

**edit → inspect → stage → inspect → commit**

| Command | Purpose |
| --- | --- |
| `git init` | Create a Git repository in the current folder |
| `git status` | Show the current state of the repository |
| `git diff` | Show changes that have not been staged |
| `git add filename` | Add a file’s changes to the staging area |
| `git diff --staged` | Show changes prepared for the next commit |
| `git commit -m "message"` | Record the staged changes |
| `git log --oneline` | Display a compact commit history |

## 6. Select, filter, and sort

Selecting one column returns a Series. Selecting several columns returns a DataFrame.

````python
coffee["revenue_usd"]
````

````python
coffee["revenue_usd"].head()
````

````python
coffee[["store", "revenue_usd", "transactions"]].head()
````

A Boolean condition creates `True` or `False` for every row. Putting that condition inside square brackets keeps only the rows where the condition is `True`. Use `&` to combine conditions, and place each condition in parentheses.

````python
coffee[coffee['store'] == 'Downtown']
````

````python
downtown = coffee[coffee["store"] == "Downtown"]
downtown.head()
````

````python
busy_downtown = coffee[
    (coffee["store"] == "Downtown")
    & (coffee["transactions"] >= 100)
]
busy_downtown.sort_values("transactions")
````

## 7. Basic summaries

Methods such as `mean()`, `median()`, `min()`, `max()`, and `sum()` reduce many values to a summary.

````python
print("Mean revenue:", coffee["revenue_usd"].mean())
print("Median revenue:", coffee["revenue_usd"].median())
print("Minimum revenue:", coffee["revenue_usd"].min())
print("Maximum revenue:", coffee["revenue_usd"].max())
print("Total revenue:", coffee["revenue_usd"].sum())
````

`groupby()` splits the rows into groups and calculates a summary for each group.

````python
store_summary = (
    coffee.groupby("store")
    .agg(
        average_revenue=("revenue_usd", "mean"),
        total_transactions=("transactions", "sum"),
    )
    .round(2)
)
store_summary
````
