## 1. What this course is about

Data Bootcamp is about using data to make better decisions and communicating the reasoning behind those decisions. Python is an important tool in that process, but code is not the final product. A strong analyst must connect four kinds of work:

1. **Frame:** Identify the stakeholder, decision, and meaningful measure of success.
2. **Analyze:** Acquire, inspect, clean, visualize, and model relevant data.
3. **Evaluate:** Check assumptions, uncertainty, limitations, ethics, and whether the work answers the original question.
4. **Communicate:** Deliver a clear recommendation that another person can understand and use.

During the semester we will move from command-line and Git workflows into Python, pandas, visualization, statistics, predictive modeling, text and time-series analysis, neural networks, and deployment. Projects will repeatedly return to the same question: **What decision will this analysis support?**


## 2. How the course works

The course follows a repeated cycle:

- **Learn:** Encounter a new idea or technique.
- **Practice:** Use it in a guided class activity and code practice.
- **Apply:** Combine techniques in a mini-project or the final project.
- **Present:** Explain the question, evidence, approach, findings, and limitations.
- **Improve:** Use feedback and reflection to revise the work.

### What you should do when stuck

Getting stuck is a normal part of data work. First, read the complete error or prompt, identify what you expected to happen, and record what you tried. Then ask a classmate, post a focused question in Slack, or attend office hours. Asking for help early is a professional skill.

## 3. What makes a data challenge business ready?

A topic is not yet a problem, and a prediction is not yet a decision. Compare these statements:

- **Topic:** customer churn
- **Analytical question:** Which customers are likely to cancel?
- **Business-ready challenge:** Which current subscribers should the retention team contact next month, what action could it take, and how will the company determine whether the intervention was worthwhile?

The final version identifies an actor, action, time frame, and need for evaluation. It gives the analysis a purpose.

A useful starting principle is:

> **Data analysis connects a stakeholder's decision to available evidence.**


### A four-part Data Challenge Canvas

| Step | Core question | A strong brief identifies |
| ---: | --- | --- |
| **1. Decision** | What problem are we solving? | The stakeholder, specific decision, deadline, available actions, constraints, and definition of success |
| **2. Data** | What evidence is available? | The unit of observation, population, variables, time period, source, information available at decision time, and important gaps |
| **3. Use** | How can the data help? | Whether the task is descriptive, predictive, causal, or an optimization; a sensible first analysis; and the claims the evidence can support |
| **4. Delivery** | How will the work lead to action? | The output the stakeholder will receive, how it will be used, major risks, and how the recommendation will be evaluated |

### Match the claim to the evidence

- **Description** summarizes or compares what was observed.
- **Prediction** estimates an unknown or future outcome.
- **Causal evaluation** asks whether an action changed an outcome and requires a credible comparison, often an experiment.
- **Optimization** recommends an action under explicit objectives and constraints.

These tasks may appear together, but they are not interchangeable. A pattern in historical data does not by itself prove what caused the pattern or what will happen after an intervention. A technically impressive answer to the wrong question is not useful.


## 4. Worked example: Staffing a coffee shop

### The initial request

> "Analyze our sales data and tell us how to improve staffing."

What is unclear? We do not yet know who will use the analysis, which staffing decision is under consideration, when the decision must be made, or what "improve" means.

### Business context

After speaking with the regional manager, we learn:

- schedules for the next month must be submitted Friday;
- customers complain about slow service during some periods; and
- store managers believe other shifts may be overstaffed.

The regional manager can change the number and roles of employees scheduled each hour, but cannot change store hours or hire additional employees this month.

### Available data

- transaction timestamp, store, product, quantity, and price for the last 12 months;
- hourly employee schedules by store and role;
- hourly labor cost by role; and
- a sample of customer wait times collected during two weeks.

### Important gaps

The transaction file records completed purchases, not customers who abandoned the line. Wait-time coverage is short and may not represent the full year. Promotions, weather, and local events are not recorded consistently. Sales can show when completed demand was high, but they cannot reveal every customer who wanted to purchase or prove that staffing caused a sales change.

### Reframed challenge

> The regional manager needs to set hourly staffing levels for three stores next month. We will compare transactions, revenue, and labor hours by store, weekday, and hour; identify recurring periods of high demand per labor hour; and use the limited wait-time sample as supporting rather than definitive evidence. We will deliver a proposed schedule adjustment and a short explanation of the evidence. Success will be evaluated with a pilot tracking labor cost, wait time, abandoned orders, and customer satisfaction.

This framing does not promise more than the data can provide. It produces an actionable recommendation and a plan to learn whether that recommendation works.


### Weak versus decision-ready framing

| Weak framing | More decision-ready framing |
| --- | --- |
| Predict house prices. | Help an acquisitions team decide which listed properties merit manual review, using only information available at listing time and reporting uncertainty. |
| Analyze employee attrition. | Help HR decide where to investigate working conditions by describing voluntary exits across comparable roles and locations, without labeling individual employees. |
| Make a sales dashboard. | Help regional managers decide where to focus next month's promotion by comparing revenue, margin, repeat purchases, and promotion history across comparable stores. |
| Find fraudulent transactions. | Help investigators prioritize a limited daily review queue while tracking false positives, financial exposure, and customer impact. |


## 5. Team workshop

Choose one scenario. Your team has **17 minutes** to draft a challenge brief and prepare a two-minute report. You may request additional information, but record unanswered questions as assumptions or limitations.

Use the time deliberately:

| Time | Task |
| ---: | --- |
| 2 minutes | Read the scenario and identify what remains ambiguous |
| 4 minutes | Define the stakeholder, decision, deadline, constraints, and success measure |
| 4 minutes | Audit the available data: unit of observation, coverage, useful fields, and gaps |
| 4 minutes | Choose the analytical task and propose the simplest useful first analysis |
| 3 minutes | Specify the deliverable, check the claim, and prepare the report |

### Scenario A: Restaurant delivery

A delivery platform receives complaints about late orders and asks you to "optimize delivery." Available tables contain order and promised times, restaurant pickup time, driver assignment and location updates, delivery time, distance, weather category, customer rating, and refund amount for six months. Driver location is sometimes missing, and canceled orders are stored separately.

### Scenario B: Consumer lending

A lender wants to "improve its approval model." Available data contains applications, stated income, requested amount, credit attributes, approval decisions, interest rates, and repayment outcomes for approved applicants. Repayment outcomes do not exist for rejected applicants. Several fields are legally or ethically sensitive.

### Scenario C: Retail promotion

A retailer asks which products should receive discounts next month. Available data contains two years of store-level daily sales, price, product cost, inventory, promotion flags, product category, store location, and holidays. Competitor prices are available for only 20% of products. Past promotions were chosen by managers rather than assigned randomly.


### Data Challenge Brief

Complete the four sections in concise sentences. Mark assumptions explicitly.

#### 1. Decision

- Who owns the decision?
- What must they decide, by when, and what actions are available?
- What constraints apply, and what business outcome defines success?

#### 2. Data

- What does one row or observation represent?
- Which population, locations, and dates are covered?
- Which fields are most relevant to the decision?
- What is missing, unreliable, selected, or unavailable at decision time?

#### 3. Use

- Is the primary task descriptive, predictive, causal, or an optimization?
- What is the simplest useful first analysis?
- What claim can the available evidence support—and what claim can it not support?

#### 4. Delivery

- What will the stakeholder receive?
- How will it change or inform the decision?
- How should the recommendation be tested, monitored, or revisited?

Summarize your brief:

> **[Stakeholder]** needs to decide **[decision]** by **[deadline]** because **[business problem]**. We have **[available data]**, where one observation represents **[unit of observation]**. We will begin by **[analysis]** and deliver **[output]** to support **[action]**. Success will be evaluated by **[criterion]**. The evidence cannot establish **[key limitation or unsupported claim]**.

### Decision-ready quality check

Before reporting, confirm that another person could answer **yes** to each question:

- Is there a named stakeholder with a real decision and feasible actions?
- Is the decision narrower than the general topic?
- Is the unit of observation clear?
- Does the proposed analysis use information available when the decision occurs?
- Does the claim match what the data and study design can support?
- Is the deliverable usable, and is there a plan to evaluate its effect?
