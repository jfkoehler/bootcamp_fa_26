## 1. What this course is about

Data Bootcamp is about using data to make better decisions and communicating the reasoning behind those decisions. Python is an important tool in that process, but code is not the final product. A strong analyst must connect four kinds of work:

1. **Frame:** Identify the stakeholder, decision, and meaningful measure of success.
2. **Analyze:** Acquire, inspect, clean, visualize, and model relevant data.
3. **Evaluate:** Check assumptions, uncertainty, limitations, ethics, and whether the work answers the original question.
4. **Communicate:** Deliver a clear recommendation that another person can understand and use.

During the semester we will move from command-line and Git workflows into Python, pandas, visualization, statistics, predictive modeling, text and time-series analysis, neural networks, and deployment. Projects will repeatedly return to the same question: **What decision will this analysis support?**

## 3. What makes a data challenge business ready?

A topic is not yet a problem, and a prediction is not yet a decision. Compare these statements:

- **Topic:** customer churn
- **Analytical question:** Which customers are likely to cancel?
- **Business-ready challenge:** Which current subscribers should the retention team contact next month, what action could it take, and how will the company determine whether the intervention was worthwhile?

The final version identifies an actor, action, time frame, and need for evaluation. It gives the analysis a purpose.

A useful starting principle is:

> **Data analysis connects a stakeholder's decision to available evidence.**

### CRISP-DM as the course roadmap

The Cross-Industry Standard Process for Data Mining (CRISP-DM) organizes data work into six connected phases:

1. **Business Understanding** — define the objective, decision, constraints, and success criteria.
2. **Data Understanding** — identify, describe, explore, and assess the available evidence.
3. **Data Preparation** — select, clean, combine, and structure the data.
4. **Modeling** — apply suitable analytical or predictive methods.
5. **Evaluation** — determine whether the results are valid and address the business objective.
6. **Deployment** — put the result into a form that people can use and monitor.

The process is iterative rather than a one-way checklist. New information discovered during data understanding may require us to revise the business question. In Class 1, we will concentrate on the first two phases and propose only a plausible next step.


### CRISP-DM Lite: the Class 1 brief

For today's exploration, reduce the framework to three questions:

| Section | Question | Include |
| --- | --- | --- |
| **Business Understanding** | What decision are we supporting? | One stakeholder, one decision, and one measure of success |
| **Data Understanding** | What evidence do we actually have? | What one row represents, the most relevant fields, and the most important limitation |
| **Analytical Direction** | What should we try first? | One feasible analysis and one useful output; label it descriptive, predictive, causal, or optimization |

A pattern in historical data can support description or prediction, but it does not by itself prove what caused the pattern. Keep the proposed analysis modest: its purpose is to establish a sensible direction, not to design the entire project before inspecting the data.


### Framework resources

CRISP-DM is our main course roadmap. These resources provide more detail or a complementary lens:

- **[IBM CRISP-DM overview](https://www.ibm.com/docs/en/spss-modeler/saas?topic=dm-crisp-help-overview):** documentation for the six-phase process used in this course.
- **[Data Science Project Scoping Guide](https://datasciencepublicpolicy.org/our-work/tools-guides/data-science-project-scoping-guide/):** a decision-first framework for connecting goals, actions, analysis, data, and ethical considerations.

Different frameworks emphasize different stages, but each discourages jumping directly from a dataset to a model.


## 4. A quick example of question–data fit

Suppose a student is interested in long waits at coffee shops and asks:

> **Would moving employees from quiet periods to busy periods reduce customer wait times without increasing labor hours?**

They find a small file containing transactions, revenue, staffing, labor hours, and sampled wait times for each store-period-date. The data is related to the question, but it covers only one week, contains missing wait times, and does not record abandoned purchases or employee roles.

**Fit assessment: partial.** The data can describe when demand, staffing, and waits were high or low. It cannot establish that moving employees would *cause* waits to fall. The student could revise the immediate question to "When and where were high wait times observed relative to transactions and labor hours?" and propose a later staffing experiment to answer the causal question.

Finding imperfect data is not failure. It helps us narrow the question, identify a proxy, request additional evidence, or explain why the original question cannot yet be answered.


## 5. Part 1 — Develop an individual question

Work independently for **10 minutes**. Begin with a problem, decision, or pattern that genuinely interests you. It may come from work, another course, your community, an industry, public policy, sports, culture, or everyday life.

Write four short responses:

1. **Problem area:** What interests or concerns you, and why?
2. **Question:** What focused question would you like evidence to answer?
3. **Use:** Who might care about the answer, and what could they do with it?
4. **Possible evidence:** What would an ideal dataset contain, and what would one row represent?

Your first question does not need to be perfect. Avoid starting with a method such as "build a model" or a deliverable such as "make a dashboard." Start with what you want to understand or decide.


## 6. Part 2 — Share and select a group question

In groups of three or four, spend **12 minutes** comparing your questions:

1. Each person has one minute to explain the problem, question, and why it matters.
2. Clarify each question without turning it into a different question.
3. Select one question for the group to investigate. You may combine questions only if the result remains focused.

Use these criteria rather than simply choosing the first idea:

| Criterion | Ask |
| --- | --- |
| **Interest** | Does the group genuinely want to know the answer? |
| **Clarity** | Could someone outside the group understand the question? |
| **Usefulness** | Who could use the answer, and for what? |
| **Feasibility** | Is relevant data likely to exist at a usable level of detail? |
| **Scope** | Could a first useful analysis be completed in a few weeks? |

Record the selected question and one reason the group chose it. An interesting question may be difficult; a straightforward question may be less meaningful. Your goal is a defensible balance, not certainty.

For the strongest candidate dataset, record its title, provider, and URL. Determine what one row represents, its population and time coverage, relevant fields, access format, and largest limitation. Then classify the fit:

- **Direct:** the data measures the central concepts at the needed level;
- **Partial or proxy:** the data addresses only part of the question or measures a related concept; or
- **Not currently answerable:** essential observations or variables are absent.

If the fit is partial or poor, revise the question to match what the data can support, identify another dataset that would be needed, or explain why the original question should remain unanswered.
