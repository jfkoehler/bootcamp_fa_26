## 1. What this course is about

Data Bootcamp is about using data to make better decisions and communicating the reasoning behind those decisions. Python is an important tool in that process, but code is not the final product. A strong analyst must connect four kinds of work:

1. **Frame:** Identify the stakeholder, decision, and meaningful measure of success.
2. **Analyze:** Acquire, inspect, clean, visualize, and model relevant data.
3. **Evaluate:** Check assumptions, uncertainty, limitations, ethics, and whether the work answers the original question.
4. **Communicate:** Deliver a clear recommendation that another person can understand and use.

During the semester we will move from command-line and Git workflows into Python, pandas, visualization, statistics, predictive modeling, text and time-series analysis, neural networks, and deployment. Projects will repeatedly return to the same question: **What decision will this analysis support?**


## 2. How the course works

The course follows a repeated cycle:

- **Learn:** Encounter a new idea or technique.
- **Practice:** Use it in a guided class activity and code practice.
- **Apply:** Combine techniques in a mini-project or the final project.

Getting stuck is a normal part of data work. First, read the complete error or prompt, identify what you expected to happen, and record what you tried. Then ask a classmate, post a focused question in Slack, or attend office hours. Asking for help early is a professional skill.

## 3. What makes a data challenge business ready?

A topic is not yet a problem, and a prediction is not yet a decision. Compare these statements:

- **Topic:** customer churn
- **Analytical question:** Which customers are likely to cancel?
- **Business-ready challenge:** Which current subscribers should the retention team contact next month, what action could it take, and how will the company determine whether the intervention was worthwhile?

The final version identifies an actor, action, time frame, and need for evaluation. It gives the analysis a purpose.

A useful starting principle is:

> **Data analysis connects a stakeholder's decision to available evidence.**

### CRISP-DM as the course roadmap

The Cross-Industry Standard Process for Data Mining (CRISP-DM) organizes data work into six connected phases:

1. **Business Understanding** — define the objective, decision, constraints, and success criteria.
2. **Data Understanding** — identify, describe, explore, and assess the available evidence.
3. **Data Preparation** — select, clean, combine, and structure the data.
4. **Modeling** — apply suitable analytical or predictive methods.
5. **Evaluation** — determine whether the results are valid and address the business objective.
6. **Deployment** — put the result into a form that people can use and monitor.

The process is iterative rather than a one-way checklist. New information discovered during data understanding may require us to revise the business question. In Class 1, we will concentrate on the first two phases and propose only a plausible next step.


### CRISP-DM Lite: the Class 1 brief

For today's exploration, reduce the framework to three questions:

| Section | Question | Include |
| --- | --- | --- |
| **Business Understanding** | What decision are we supporting? | One stakeholder, one decision, and one measure of success |
| **Data Understanding** | What evidence do we actually have? | What one row represents, the most relevant fields, and the most important limitation |
| **Analytical Direction** | What should we try first? | One feasible analysis and one useful output; label it descriptive, predictive, causal, or optimization |

A pattern in historical data can support description or prediction, but it does not by itself prove what caused the pattern. Keep the proposed analysis modest: its purpose is to establish a sensible direction, not to design the entire project before inspecting the data.


## 4. Worked example: Staffing a coffee shop

### The initial request

> "Analyze our sales data and tell us how to improve staffing."

What is unclear? We do not yet know who will use the analysis, which staffing decision is under consideration, when the decision must be made, or what "improve" means.

### Business context

After speaking with the regional manager, we learn:

- schedules for the next month must be submitted Friday;
- customers complain about slow service during some periods; and
- store managers believe other shifts may be overstaffed.

The regional manager can change the number and roles of employees scheduled each hour, but cannot change store hours or hire additional employees this month.

### Available data

For this example, we will use the **[synthetic coffee-shop operations dataset](data/coffee_shop_operations_synthetic.csv)** included with the course book. It is fictional but structured like an operating report. Each row represents one store during one service period on one date. It contains:

- transaction timestamp, store, product, quantity, and price for the last 12 months;
- hourly employee schedules by store and role;
- hourly labor cost by role; and
- a sample of customer wait times collected during two weeks.

### Important gaps

The transaction file records completed purchases, not customers who abandoned the line. Wait-time coverage is short and may not represent the full year. Promotions, weather, and local events are not recorded consistently. Sales can show when completed demand was high, but they cannot reveal every customer who wanted to purchase or prove that staffing caused a sales change.

### CRISP-DM Lite brief

- **Business Understanding:** The regional manager must reallocate existing labor hours across store service periods for next month. Success means reducing wait times without increasing total labor hours.
- **Data Understanding:** Each row represents one store-period-date. Transactions, revenue, staffing, and labor hours are useful, but only one week is covered and abandoned purchases are unobserved.
- **Analytical Direction:** Begin with a descriptive comparison of demand, wait time, and transactions per labor hour across stores and periods. Deliver a small set of proposed schedule changes to evaluate in a pilot.

The brief is intentionally short. It establishes a direction without pretending the first analysis will settle the staffing decision.


### Weak versus decision-ready framing

| Weak framing | More decision-ready framing |
| --- | --- |
| Predict house prices. | Help an acquisitions team decide which listed properties merit manual review, using only information available at listing time and reporting uncertainty. |
| Analyze employee attrition. | Help HR decide where to investigate working conditions by describing voluntary exits across comparable roles and locations, without labeling individual employees. |
| Make a sales dashboard. | Help regional managers decide where to focus next month's promotion by comparing revenue, margin, repeat purchases, and promotion history across comparable stores. |
| Find fraudulent transactions. | Help investigators prioritize a limited daily review queue while tracking false positives, financial exposure, and customer impact. |


## 5. Team workshop

Choose one scenario. Your team has **20 minutes** to draft a challenge brief and prepare a two-minute report. You may request additional information, but record unanswered questions as assumptions or limitations.

Use the time deliberately:

| Time | Task |
| ---: | --- |
| 2 minutes | Read the scenario and identify what remains ambiguous |
| 5 minutes | Complete Business Understanding: one stakeholder, decision, and success measure |
| 5 minutes | Complete Data Understanding: one row, useful fields, and the largest limitation |
| 3 minutes | Propose one analytical direction and useful output |
| 2 minutes | Check the claim and prepare the report |

### Scenario A: Federal Reserve data and business planning

A mid-sized home-furnishings retailer is preparing its first-quarter 2027 budget. The CFO asks your team to "use Federal Reserve data to tell us what the economy will do" so that the company can set borrowing, inventory, and pricing assumptions.

Use the **[FRED business-conditions dataset](data/fred_business_conditions.csv)**. Each row represents one month and contains the effective federal funds rate (`FEDFUNDS`), Consumer Price Index (`CPIAUCSL`), and unemployment rate (`UNRATE`). The series begin on different dates, the most recent observations may be incomplete or revised, and the file contains no company sales, costs, customers, or inventory. CPI is a price index; a change in the index, rather than its level alone, is needed to discuss inflation.

**Source:** [Federal Reserve Economic Data](https://fred.stlouisfed.org/) at the Federal Reserve Bank of St. Louis. The CPI and unemployment measures originate with the U.S. Bureau of Labor Statistics.

### Scenario B: UCI Bank Marketing

A Portuguese bank's marketing director wants to decide which existing clients should receive calls about a term deposit in the next campaign. The director asks your team to "build a model that maximizes subscriptions" while keeping the number of calls within the team's capacity.

Use the **[UCI bank-marketing sample](data/uci_bank_marketing_sample.csv)**. Its 4,521 rows are a random 10% sample from 45,211 marketing examples collected from May 2008 through November 2010. The 16 inputs describe client attributes and current or previous campaign contacts; `y` records whether the client subscribed. The file uses semicolons as separators. The `duration` field is known only after the most recent call ends, so it is not valid information for deciding whom to call beforehand. Historical contacts were not described as a randomized experiment, and fields such as age, job, marital status, credit default, and loan status require careful ethical review.

**Source:** [UCI Machine Learning Repository: Bank Marketing](https://archive.ics.uci.edu/dataset/222/bank+marketing), created by Paulo Cortez and Sérgio Moro.

### Scenario C: Coffee-shop operations

A regional manager must set next month's staffing levels for two coffee shops without increasing total scheduled labor hours. The manager asks your team to identify where labor should be added or removed.

Use the **[synthetic coffee-shop operations dataset](data/coffee_shop_operations_synthetic.csv)**. Its 42 rows cover two stores, three service periods, and seven days. Each row contains completed transactions, revenue, scheduled staff, labor hours, a sampled average wait time, promotion status, and rain status for one store-period-date. Some wait-time values are missing. The dataset contains only one week and does not record abandoned purchases, employee roles, worker availability, customer satisfaction, or variation within a service period. Because the data is synthetic, it can support classroom reasoning but not a claim about a real business.


### CRISP-DM Lite Challenge Brief

Write no more than three bullets:

- **Business Understanding:** **[Stakeholder]** must decide **[specific decision]**. Success means **[business outcome]**.
- **Data Understanding:** One row represents **[unit of observation]**. The most useful fields are **[fields]**, but **[critical limitation]**.
- **Analytical Direction:** Begin with **[descriptive, predictive, causal, or optimization task]** by **[first analysis]** and provide **[output]**.

Before reporting, check three things: the decision is actionable, the unit of observation is correct, and the proposed claim does not exceed the evidence.
