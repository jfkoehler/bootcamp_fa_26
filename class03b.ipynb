# git and introductory `pandas`

Continuing on our work with the terminal and UNIX last class, we will explore a basic `git` workflow and introduce the pandas library.

## Learning Objectives

By the end of class you should be able to:

- create a `git` repository, inspect changes, stage a file, commit it, synchronize it with github
- understand the difference between the four basic Python data types -- string, int, float, and bool
- explain the differences between built in Python data structures -- list, tuple, dict, set
- use the `pandas` library to create a DataFrame from a .csv file, select columns, filter rows, perform basic numerical summaries
